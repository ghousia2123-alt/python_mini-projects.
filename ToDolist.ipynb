```python
# Simple To-Do List

tasks = []

while True:
    print("\n1. Add Task")
    print("2. View Tasks")
    print("3. Delete Task")
    print("4. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        task = input("Enter task: ")
        tasks.append(task)
        print("Task added")

    elif choice == 2:
        print("\nYour Tasks:")
        for task in tasks:
            print("-", task)

    elif choice == 3:
        task = input("Enter task to delete: ")

        if task in tasks:
            tasks.remove(task)
            print("Task deleted")
        else:
            print("Task not found")

    elif choice == 4:
        print("Thank you")
        break

    else:
        print("Invalid choice")
```
